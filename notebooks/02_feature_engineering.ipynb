# 02 - Feature Engineering
We build 6 behaviour-based features from the raw bill/payment history and check that they relate to default.
The real work is done by `src/feature_engineering.py`; the saved train/test files are produced by `python -m src.data_preprocessing`.

````python
import os, sys
sys.path.insert(0, os.path.abspath(".."))   # so we can import from src/

import pandas as pd
import matplotlib.pyplot as plt
from src.data_preprocessing import load_raw_data, clean_data
from src.feature_engineering import add_features
````

````python
df = clean_data(load_raw_data())
df_fe = add_features(df)

new_cols = ["avg_repayment_delay", "max_repayment_delay", "avg_bill_amt",
            "avg_payment_amt", "payment_to_bill_ratio", "credit_utilization"]
df_fe[new_cols].describe().T
````

### Do the new features relate to default?
A quick correlation with the target. Higher repayment delay should go with higher default.

````python
corr = df_fe[new_cols + ["default"]].corr()["default"].drop("default").sort_values()
print(corr.round(3))

corr.plot(kind="barh", title="Correlation of new features with default")
plt.show()
````

### Check for bad values
Infinity / missing values would break model training. The pipeline handles them; here we just confirm.

````python
import numpy as np
print("Infinite values:", np.isinf(df_fe[new_cols]).sum().sum())
print("Missing values :", df_fe[new_cols].isna().sum().sum())
````
